```python
import requests
import json
import time

API_URL = "https://api.spyne.ai/360/v1/reprocess"

HEADERS = {
    "authority": "api.spyne.ai",
    "accept": "*/*",
    "accept-language": "en-US",
    "content-type": "application/json",
    "sec-ch-ua": '" Not A;Brand";v="99", "Chromium";v="102"',
    "sec-ch-ua-mobile": "?0",
    "sec-ch-ua-platform": "Windows",
    "sec-fetch-dest": "empty",
    "sec-fetch-mode": "cors",
    "sec-fetch-site": "cross-site",
    "user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "spyne/6.2.0 Chrome/102.0.5005.167 Electron/19.1.9 Safari/537.36"
}

USER_EMAIL = "jatin@spyne.ai"

SKU_BG_MAP = {
  "81025348e9914b4db1b5c7d295c86897":545454
}

for sku_id, bg_id in SKU_BG_MAP.items():
    payload = {
        "studio_classifier": "0",
        "stablize": "1",
        "reason": "reprocess",
        "sku_id": sku_id,
        "bgId": bg_id,
        "user_email_id": USER_EMAIL
    }

    print(f"Reprocessing SKU: {sku_id} | bgId: {bg_id}")

    try:
        response = requests.post(
            API_URL,
            headers=HEADERS,
            data=json.dumps(payload),
            timeout=30
        )

        print(f"Status: {response.status_code}")
        print(f"Response: {response.text}")

    except Exception as e:
        print(f"❌ Error for SKU {sku_id}: {str(e)}")

    print("-" * 60)
    time.sleep(0.5)  # avoid rate limiting
```
