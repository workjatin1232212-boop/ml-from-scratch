```python
#CODING CHALLENGE
import pandas as pd

data = {
    "timestamp": [
        "2024-01-01 10:00", "2024-01-01 10:05", "2024-01-01 10:10",
        "2024-01-01 10:15", "2024-01-01 10:20", "2024-01-01 10:25"
    ],
    "endpoint": ["/login", "/login", "/home", "/home", "/search", "/search"],
    "response_time": [120, 250, 180, None, 400, 150],
    "status_code": [200, 500, 200, 200, 200, 500]
}

df = pd.DataFrame(data)
```

```python
from datetime import datetime
df['timestamp']=pd.to_datetime(df['timestamp'])
df
```

```python
df['response_time'].isna().sum()
```

```python
df['response_time']=df['response_time'].fillna(df['response_time'].mean())
```

```python
df
```

```python
df=df.drop_duplicates()
```

```python
df
```

```python
df['endpoint'].unique()
```

```python
df.groupby('endpoint').agg(avg_response_time=('response_time','mean'))
```

```python
checkStatusCode=df[df['status_code']>= 400]
checkStatusCode
```

```python
slowest_endpoit=df.groupby('endpoint').agg(min_res_time=('response_time','mean'))
print(slowest_endpoit.max())

```
